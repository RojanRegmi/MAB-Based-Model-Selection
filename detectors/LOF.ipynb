```python
from sklearn.neighbors import LocalOutlierFactor
from sklearn.preprocessing import MinMaxScaler

import pandas as pd
import numpy as np

import os
import pickle

from sklearn.metrics import precision_score, recall_score, f1_score
import matplotlib.pyplot as plt 
%matplotlib inline

import sys
sys.path.append('..')

from src.utils import train_test_anomaly, raw_thresholds
```

```python
file_path = '../datasets/Dodgers/101-freeway-traffic.test.out'

columns = ['value', 'anomaly']

df = pd.read_csv(file_path, names=columns, header=None)
```

```python

min_max_scaler = MinMaxScaler()
scaled_data = min_max_scaler.fit_transform(df[['value']])
```

```python
df = df[df['value'] >= 0]
df['anomaly'].value_counts()
```

```python
model = LocalOutlierFactor(n_neighbors=50, contamination=0.057)
thres_np = model.fit_predict(scaled_data)
thres_np = (thres_np + 2) % 3
```

```python
unique_values, counts = np.unique(thres_np, return_counts=True)
value_counts = dict(zip(unique_values, counts))
print("Value counts:", value_counts)
```

```python
print(np.min(thres_np))
print(np.max(thres_np))
```

```python
gtruth_np = df[['anomaly']].values.reshape(-1, 1)
gtruth_flat = gtruth_np.flatten()
gtruth_flat.shape
```

```python
prec = precision_score(gtruth_flat, thres_np)
recall = recall_score(gtruth_flat, thres_np)
f1 = f1_score(gtruth_flat, thres_np)
```

```python
print(f'Precsion Score: {prec:.4f}  Recall: {recall:.4f}  f1_score: {f1:.4f}')
```
